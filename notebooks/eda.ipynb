# EDA — synthetic demo dataset (GoNB)

Exploratory scratch for `data/raw/sample.csv`, written for the [GoNB](https://github.com/janpfeifer/gonb) Jupyter kernel for Go. Notebooks are scratch, not production code (research §3.3): anything that proves useful graduates into `internal/`.

Run with: `jupyter lab` using the GoNB kernel (`gonb --install`).

```go
import (
    "encoding/csv"
    "fmt"
    "os"
)

f, _ := os.Open("../data/raw/sample.csv")
rows, _ := csv.NewReader(f).ReadAll()
f.Close()
fmt.Println("header:", rows[0])
fmt.Println("records:", len(rows)-1)
counts := map[string]int{}
for _, r := range rows[1:] { counts[r[len(r)-1]]++ }
fmt.Println("per class:", counts)
```
